```python
import numpy as np
```

```python
# columns: [age, math mark , science mark]
data = np.array([
                [18,85,78],  # student 1
                [19,92,88],  # student 2
                [17,76,95],  # student 3
                [18,65,70],  # student 4
                [20,90,85]   # student 5
                         ])
```

```python
data.shape
```

```python
# find the avarage of age students
np.mean(data[:,0])
```

```python
# extract maths mark of the student 
data[:,0]
```

```python
# find the higest science mark
np.max(data[:,2])
```

```python
# get details of the student who scored mored than 90 in maths
data[data[:,1] > 90]
```

```python
# increase maths mark all student by 5
data[:,1] += 5
data
```

```python
# find student who yeanger than 19
len(data[data[:,0] < 19])
```

```python
# calculate the avarage marks in each subject (columns-wise  mean)
np.mean(data[:,1:],axis = 0)
```

```python
# get data of students who at least scored 80 marks in both subject
data[(data[:,1] >= 80 ) & (data[:,2] >= 80)]
```

```python
# replace all science marks < 75 with zero
data[:,2][data[:,2] < 75 ] = 0
data
```
